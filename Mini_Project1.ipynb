```python
# Student Report Card Generator

students = [
    {"name": "Ishan", "subjects": {"Math": 85, "Science": 90, "English": 80}, "is_active": True},
    {"name": "Gautam", "subjects": {"Math": 95, "Science": 88, "English": 92}, "is_active": True},
    {"name": "Manas", "subjects": {"Math": 70, "Science": 65, "English": 60}, "is_active": False},
    {"name": "Aditi", "subjects": {"Math": 88, "Science": 75, "English": 70}, "is_active": True},
    {"name": "Khushi", "subjects": {"Math": 50, "Science": 55, "English": 70}, "is_active": False}
]

def report_card(student):

    print("\n===========================")
    print("   STUDENT REPORT CARD")
    print("===========================")
    print("Name :", student["name"])

    if student["is_active"]:

        total = sum(student["subjects"].values())
        average = total / len(student["subjects"])

        if average >= 90:
            grade = "A+"
        elif average >= 75:
            grade = "A"
        elif average >= 60:
            grade = "B"
        else:
            grade = "C"

        if average >= 60:
            status = "PASS"
        else:
            status = "FAIL"

        print("Average :", f"{average: .2f}")
        print("Grade   :", grade)
        print("Status  :", status)
        print("Active  :", student["is_active"])

    else:
        print("Status  : Inactive Student")
        print("Marks are Hidden")

    print("===========================")


print("Student Report Card Generator")

for student in students:
    report_card(student)
```
